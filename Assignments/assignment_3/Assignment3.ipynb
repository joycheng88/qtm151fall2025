(b) Operations on Arrays <font size = "3"> 

- Define an array `temps_c` using `np.array()`, assigning the following values of temperatures measured in degrees Celsius: <br> <br>`temps_c` $= \begin{pmatrix} c_1 \\ c_2 \\ c_3 \\ c_4 \\ c_5 \end{pmatrix} = \begin{pmatrix} 0 \\ 10 \\ 20 \\ 30 \\ 40 \end{pmatrix}$ <br> <br>
- Use element-by-element vector operations to compute a new array `temps_f` of length 5, which will convert the temperatures from Celsius to Fahrenheit: <br><br>
`temps_f` $= \begin{pmatrix} f_1 \\[8pt] f_2 \\[8pt] f_3 \\[8pt] f_4 \\[8pt] f_5 \end{pmatrix} = \begin{pmatrix} c_1 \times \tfrac{9}{5} +32 \\[8pt] c_2 \times \tfrac{9}{5} +32 \\[8pt] c_3 \times \tfrac{9}{5} + 32 \\[8pt] c_4 \times \tfrac{9}{5} + 32 \\[8pt] c_5 \times \tfrac{9}{5} + 32\end{pmatrix} $ <br><br>

- To get full points you must do all of the above tasks


```python
# Write your answer here

# Define an array of Celsius temperatures
temps_c = np.array([0, 10, 20, 30, 40])

# Compute an array of Fahrenheit temperatures
temps_f = (temps_c * 9/5) + 32

# Print the result
print("The array of Fahrenheit temperatures are:", temps_f)

# Compute the mean of Farenheit temperatures
temps_f.mean()

# Print the result
print("The mean of Farenheit temperatures is:", temps_f.mean())
```

(c) Plot a Normal Random Variable
<font size = "3">

To get full points:

- Set the seed using `np.random.seed(1234)`
- Make the random variable with the following parameters:
    -  mean = 7
    - standard deviation = 5
    - size = 1500

- Make a histogram
- Create a title
- Label the axes

```python
# Write your answer here
np.random.seed(1234)

# Define parameters
mu = 7 # mean
sd = 5 # standard deviation
n = 1500 # size of sample

# Generate random sample in uniform distribution
x = np.random.uniform(mu, sd, n)

# Make a histogram
plt.hist(x, bins=30, edgecolor='black')

# Add title and labels
plt.title("Histogram of Normal Random Variable")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

(e) Classifying a Variable

<font size = "3">

- Below, the variable `pizza_rating` is assigned a random value between 0 and 10
- Write an if/elif/else statement that does the following
  - `print("Best Pizza Ever")` if the rating is **greater than or equal to 9**.  
  - `print("Pretty Good")` if the rating is **greater than or equal to 7 but less than 9**.  
  - `print("Meh")` if the rating is **greater than or equal to 4 but less than 7**.  
  - `print("Never Again")` otherwise.

- To get full points you must do all of the above tasks 

```python
pizza_rating = np.random.uniform(low = 0, high = 10)

# Write your answer here
# if/elif/else statement for pizza_rating
if pizza_rating >= 9:
    print("Best Pizza Ever")
elif pizza_rating >= 7: 
    print("Pretty Good")
elif pizza_rating >= 4:
    print("Meh")
else:
    print("Never Again")

# Check value of pizza_rating
print("The pizza rating is:", pizza_rating)
```
