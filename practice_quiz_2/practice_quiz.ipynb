(d) Filtering a data set
<br>

<font size = "3">

Create two DataFrames containing the following subsets of the data:

- People who weight satisfies $115 \leq \textrm{weight} \leq 140$ (in pounds)
- People who either i) weigh strictly less than 95 pounds **or** ii) are 72 inches or taller (greater than or equal to 72 inches)

```python
# Write your answer here:
# People whose weight is between 115 and 140 
subset1_hw = df_hw[(df_hw['Weight_Pounds'] >= 115) & (df_hw['Weight_Pounds'] <= 140)]
display(subset1_hw.head())

# People who weigh strictly less than 95 or are at least 72 inches tall
subset2_hw = df_hw[(df_hw['Weight_Pounds'] < 95) | (df_hw['Height_Inches'] >= 72)]
display(subset2_hw.head())
```

(e) Random Assignment

<font size = "3">

We want to divide the dataset into 3 groups randomly: group A, group B, group C

Randomly assign each person in the dataset one of three possible strings: ``"A"``, ``"B"``, or ``"C"``. Add a column/variable to the DataFrame which contains the letter each person is assigned. Give the column/variable the name ``"Group"``

**Hint**: Randomly choosing among 3 strings is not much different from randomly choosing among 4 numbers, which is what we did in Lecture 11

```python
# Write your answer here: 
# Randomly assign each observation "A", "B", or "C" 
group_letter = np.random.choice(["A", "B", "C"], size = df_hw.shape[0])

# Add a column and set the column name as "Group"
df_hw["Group"] = group_letter

display(df_hw.head())
```

(f) Plotting data from one the groups

<font size = "3">

Create a new DataFrame by extracting all rows corresponding to Group B. Then make a scatter plot of height vs. weight (height on horizontal axis, weight on vertical axis) for the people in this group.

Give appropriate labels to the axes and give your scatter plot a reasonable title.

```python
# Write your answer here:
subset_B = df_hw[(df_hw["Group"] == "B")]
display(subset_B.head())

# Make a scatter plot of height vs. weight for people in subset_B 
plt.figure()
plt.scatter(x = subset_B["Height_Inches"], y = subset_B["Weight_Pounds"], alpha = 0.4)
plt.xlabel("Height (inches)")
plt.ylabel("Weight (pounds)")
plt.title("Height vs. Weight for Subset B")
plt.show()
```
