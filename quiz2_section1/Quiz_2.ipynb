(4) Create a function and apply it to a column <br>

<font size = "3">

Step 1:
- Define a function with a single input argument "TotalStats" which classifies each player as follows:
    - If "TotalStats > 85", the function returns the string "Top Performer"
    - If "TotalStats <= 85", the function returns the string "Non Top-Performer"

Step 2:
- Extract the column "TotalStats" from the FIFA players DataFrame, and assign it to a variable (which should be a Pandas Series) <br>

- Using the Pandas ``.apply`` method, create a new Pandas Series called "player_classification" which contains the appropriate string for each player.

````python
# Write your own code
# Define a function that classifies each player as "Top Performer" or "Non Top-Performer"
def totalStatsClassifier(Overall):
    if Overall > 85: 
        return "Top Performer"
    else:
        return "Non Top-Performer"

# Extract teh column "Overall" from the dataframe df_fifa
overall = df_fifa["Overall"]
print(type(overall))  # Pandas Series
display(overall.head())

# Apply the function totalStatsClassifier to each element in overall Series
player_classification = overall.apply(totalStatsClassifier)
print(type(player_classification))  # Pandas Series
display(player_classification)

# Check how many players are classified as "Top Performer" and "Non Top-Performer"
player_count = player_classification.value_counts()
display(player_count)
````

(5) Split a dataset into subsets
 

<font size = "3">

- We will continue using the DataFrame you read in from "fifa23_players.csv"
- Use ``.shape`` to define an integer variable called ``n``, which is the total number of rows in this DataFrame. <br>
- Use ``numpy.random.uniform`` to create a random sample of size ``n`` from the uniform distribution between 0 and 1. Then store these random numbers as a new column ``random_var`` to the original DataFrame.<br> 
- Use ``.query`` to extract all players with ```random_var < 0.4```, and assign it to a new DataFrame called ``groupA``
- Use ``.query`` to extract all players with ```random_var >= 0.4```, and assign it to a new DataFrame called ``groupB``.

**Note**: In this way, we are assigning roughly 40% of the players to Group A, and roughly 60% to Group B


````python
# Write your own code
# Define n as the number of players (total rows) in the dataset
n = df_fifa.shape[0]
print("Total number of players:", n) 

# Assign a random value between 0 and 1 to each player
random_var = np.random.uniform(0, 1, n) # Uniform distribution on [0,1]

# Create a new column "random_var" to store the random values
df_fifa["random_var"] = random_var
display(df_fifa.head())

# Split the dataset into two groups (40%, 60%) based on the random values
groupA = df_fifa.query("random_var < 0.4")
groupB = df_fifa.query("random_var >= 0.4")
print("Group A shape:", groupA.shape)
print("Group B shape:", groupB.shape)
````

(6) Create two scatter plots of the groups

<font size = "3">

- This question requires the two datasets you created in part (5)
- Use ``matplotlib.pyplot.subplots`` to create two side by side scatter plots (1 row, 2 columns)
- In the left figure, plot the columns ``Age`` vs. ``Overall`` for the data in ``groupA`` (so the horizontal axis will correspond to "Age")
- In the right figure, plot the columns ``Age`` vs. ``Overall`` for the data in ``groupB`` (so the horizontal axis will correspond to "Age")
- Label the axes for both plots. Create a title for **both** plots, indicating whether Group A or Group B is shown.

**Note**: If the assignment into groups was done randomly, then we expect that the scatter plots will be **very similar**

````python
# Your code here
# Create subplots side by side
fig, axes = plt.subplots(1, 2, figsize=(10, 5))

# Scatter plot for Group A
# Horizontal axis is Age, Vertical axis is Overall
axes[0].scatter(groupA['Age'], groupA['Overall'], color='blue')
axes[0].set_title('Group A')
axes[0].set_xlabel('Age')
axes[0].set_ylabel('Overall')

# Scatter plot for Group B
axes[1].scatter(groupB['Age'], groupB['Overall'], color='red')
axes[1].set_title('Group B')
axes[1].set_xlabel('Age')
axes[1].set_ylabel('Overall')

plt.suptitle('Scatter Plot: Age vs. Overall for Two Groups of Players')
plt.xlabel('Age')
plt.ylabel('Overall')
plt.tight_layout()
plt.show()
````
