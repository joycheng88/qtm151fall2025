```python
# Check duplicates
# have duplicates -> print true, easier to find counterexample
# no duplicates -> print false, need to check each element
list1 = [1, 2, 3, 2, 5, 6, 6, 9]
print(set(list1))

# Method 1: using set
if len(list1) != len(set(list1)):
    print(True)
else: 
    print(False)

# Method 2: for-else structure
for i in range(len(list1)):
    list1[i]
    if list1[i] in list1[:1] + list1[i+1:]: # Rest of the list
        print(True)
        break # need only one counterexample
else: 
    print(False)

# Method 3: using new list
new_list = []
for i in list1: 
    if i not in new_list:
        new_list.append(i)
    else: 
        print(True)
        break
else: 
    print(False)
```

```python
# Find intersection of 2 lists
list1 = [1, 2, 3]
list2 = [2, 3, 4]

# & - intersection; | for union 
intersection = list(set(list1) & set(list2))  # Using set intersection
print(intersection)  # [2, 3]

intersection1 = []
for i in list1 and i not in intersection1: 
    if i in list2: 
        intersection1.append(i)
print(intersection1)  # [2, 3]
```

```python
# Find union of 2 lists
union = list(set(list1) | set(list2))  # Using set union
print(union)  # [1, 2, 3, 4]

union1 = []
for i in list1 + list2: 
    if i not in union1:
        union1.append(i)
print(union1)  # [1, 2, 3, 4]
```

```python
# Sorting
numbers = [1, 20, 4, 3, 7, 4, 8]

# Bubble sort 冒泡排序算法
print(sorted(numbers))

for i in range(len(numbers)): # 控制范围
    for j in range(0, len(numbers)-i-1): # 比较的index
        if numbers[j] > numbers[j+1]:
            numbers[j], numbers[j+1] = numbers[j+1], numbers[j] # Change order
print(numbers)
```

```python
# Dictionary algorithm
# count characters in a string
str1 = "hello world"
char_freq = {}
for char in str1:
    if char in char_freq:
        char_freq[char] += 1
    else: 
        char_freq[char] = 1
print(char_freq)  # {'h': 1, 'e': 1, 'l': 3, 'o': 2, ' ': 1, 'w': 1, 'r': 1, 'd': 1}
```

```python
# Find all prime numbers in a list
# argument: list
# return: new list of primes

def is_prime(n): 
    factor = 2
    while factor <= n**0.5: 
        if n % factor == 0: 
            return False
        factor += 1
    else: 
        return True

def all_prime(lis): 
    prime_list = []
    for i in lis: 
        if is_prime(i): 
            prime_list.append(i)
    return prime_list

# generate a list of random number (non-repetitive)
def generate_random_list(n): 
    lis = []
    while len(lis) < n: 
        num = random.randint(1, 500)
        if num not in lis: 
            lis.append(num)
    return lis

import random
random_list = generate_random_list(10)
print(random_list)

prime_list = all_prime(random_list)
print(prime_list)

# Generate consecutive integers (out-of-order)
def generate_consecutive_list(n): 
    num = list(range(1, n+1))
    random.shuffle(num)
    return num

consecutive_list = generate_consecutive_list(10)
print(consecutive_list)
```

```python
# Recursion 
def get_roots(a, b, c): 
    root1 = (-b + (b ** 2 - 4 * a * c) ** 5) / (2 * a) 
    root2 = (-b - (b ** 2 - 4 * a * c) ** 5) / (2 * a)
    return root1, root2 

r1, r2 = get_roots(1, -2, 1)
print(r1, r2)
```

```python
# Fibonacci sequence
def fibonacci(n): 
    # base case
    if (n == 1 or n == 2): 
        return 1
    # recursion call
    return fibonacci(n - 1) + fibonacci(n - 2)

fibonacci(5)
fibonacci(8)
```

```python
class Person: 
    def __init__(self, name, age, gender): 
        self.name = name
        self.age = age
        self.gender = gender
    
    def speak(self): 
        return f"Hello, everyone. My name is {self.name} and I'm {self.age} years old."
    
person1 = Person("Zhangsan", 15, "male")
person2 = Person("Lisi", 18, "female")
print(person1.speak())
print(person2.speak())

# Subclass of person: student

class Student(Person): 
    def __init__(self, name, age, gender, GPA): 
        # Super function - call parent class
        super().__init__(name, age, gender)
        self.GPA = GPA
    # Method Overriding 方法重写
    def speak(self): 
        return super().speak() + f" I'm a student, and my GPA is {self.GPA}."


student1 = Student("Alice", 15, "female", 4.0)
print(student1.speak())
```

```python
class BankAccount: 
    # Constructor
    def __init__(self, account_holder, balance = 0): 
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount): 
        if amount > 0: 
            self.balance += amount
            return f"Deposited ${amount}. New balance: ${self.balance}"
        else: 
            return "Invalid deposit amount."
        
    def withdraw(self, amount): 
        if 0 < amount <= self.balance: 
            self.balance -= amount
            return f"Withdraw ${amount}. New Balance: ${self.balance}"
        else: 
            return "Invalid withdraw amount."
        
    def get_balance(self): 
        return f"Account balance for {self.account_holder}: ${self.balance}"

b1 = BankAccount("Zhangsan")
print(b1.balance)

print(b1.deposit(5000))
print(b1.deposit(1000))
print(b1.withdraw(50))
print(b1.get_balance())

b2 = BankAccount("Lisi", 1000)
print(b2.get_balance())
```

```python
class SavingsAccount(BankAccount): 
    def __init__(self, account_holder, balance = 0, interest_rate = 0): 
        super().__init__(account_holder, balance)
        self.interest_rate = interest_rate

    def add_interest(self): 
        interest = self.balance * self.interest_rate
        self.balance += interest
        return f"Added interest. New balance: ${self.balance}"
    
s1 = SavingsAccount("Zhangsan", 500, 0.02)
print(s1.add_interest())

class CheckingAccount(BankAccount): 
    def __init__(self, account_holder, balance = 0, overdraft_limit = 100): 
        super().__init__(account_holder, balance)
        self.overdraft_limit = overdraft_limit

    def withdraw(self, amount): 
        if 0 < amount <= self.balance + self.overdraft_limit: 
            self.balance -= amount
            return f"Withdraw ${amount}. New Balance: ${self.balance}"
        else: 
            return "Invalid withdraw amount."

c1 = CheckingAccount("Lisi", 1000, 200)
print(c1.withdraw(1100))
```
