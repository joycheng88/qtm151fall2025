````python
# Write your answer
# Countries selected: Russia and India
russia_emissions = pd.read_csv('russian_federation_co2_emissions.csv')
india_emissions = pd.read_csv('india_co2_emissions.csv')

# Scatter plot for Russia
plt.figure(figsize=(8,5))
plt.scatter(data = russia_emissions, x = "year", y = "total_emissions", color = "red")
plt.title("CO2 Emissions of the Russian Federation")
plt.xlabel("Year")
plt.ylabel("Total Emissions (kt)")
plt.show()
# Russia: Total CO2 emissions dropped significantly in 1990, perhaps due to emission policy changes 
# and quota restriction.

# Scatter plot for India 
plt.figure(figsize=(8,5))
plt.scatter(data = india_emissions, x = "year", y = "total_emissions", color = "blue")
plt.title("CO2 Emissions of India")
plt.xlabel("Year")
plt.ylabel("Total Emissions (kt)")
plt.show()
# India: Total CO2 emissions increased steadily from 1950 to 2010, likely due to rapid industrialization
````
