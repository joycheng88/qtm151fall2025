- Read in the dataset named "features.csv".
- Create a ``` lambda ``` function that checks if weight $≤$ 3000
- Using the Pandas method ``.apply()``, create a new Series that shows ``True`` if weight $≤$ 3000 and ``False``if weight > 3000.
- Using the ``.shape`` attribute, print the number of cars that have weight $≤$ 3000.

**Hint**: A Pandas Series has only one dimension, so the ``.shape`` attribute is a "tuple" with only one element. In order to get a single integer, type ``your_variable_name.shape[0]``


````python
# Write your answer here:
# Load the dataset "features.csv"
df_features = pd.read_csv("data/features.csv")
display(df_features.head())

# Create a lambda function to check if weight <= 3000
check_weight = lambda weight: weight <= 3000 

# Create a new Series with True/False values
weight_condition = df_features["weight"].apply(check_weight)
display(weight_condition.head())

# Print the number of cars that have weight <= 3000
num_cars = weight_condition[weight_condition == True].shape[0]
print("Number of cars with weight less than or equal to 3000:", num_cars)
````

(b) Subsetting Data

<font size = "3" >

- Read in the dataset named "features.csv".
- Define a global variable called ```threshold``` that takes the value 2500 <br>
- Subset the data to get one dataset with weight $>$ threshold and <br>
another with weight $\le$ threshold. Use the ```.query()``` function.
- Compute the mean acceleration for each of these datasets, printing these values to the screen<br>

Note: Review how to define a global variable inside ```.query()```
 

````python
# Write your answer here:
threshold = 2500

df_heavy_cars = df_features.query("weight > @threshold")
df_light_cars = df_features.query("weight <= @threshold")

# Compute mean acceleration for each dataset
mean_acc_heavy = df_heavy_cars["acceleration"].mean()
mean_acc_light = df_light_cars["acceleration"].mean()

# Print the results
print(f"Mean acceleration for cars with weight > threshold: {mean_acc_heavy:.2f}")
print(f"Mean acceleration for cars with weight <= threshold: {mean_acc_light:.2f}")
````

<font size = "3">

Read in the dataset named "wdi_urban.csv".

  This contains data from 1980 and 2020.

- **Create an array with the unique values of "year" using the ``Pandas`` function ``unique``**

- **Create a plot with a ...**
    - Histogram of "prop_urbanpopulation" (proportion of urban population) for each year
    - Set the histogram option `alpha = 0.5` for transparency
    - All the histograms should be overlapping on the same plot
    - Create a legend with the list of years
    - Label the axes and the title

**Hint**: The procedure will be similar (but not exactly the same!) as when we created a scatter plot of cars with different numbers of cylinders in Lecture 11

````python
# your answer here
df_wdi_urban = pd.read_csv("data/wdi_urban.csv")
display(df_wdi_urban.head())

years = df_wdi_urban["year"].unique()

# Create a histogram of prop_urbanpopulation for each year
plt.figure()
for year in years:
    df_year_subset = df_wdi_urban.query("year == @year")["prop_urbanpopulation"]
    plt.hist(df_year_subset, alpha=0.5, label=str(year))

# Add legend, labels, and title
plt.legend(title="Year")
plt.xlabel("Proportion of Urban Population")
plt.ylabel("Frequency")
plt.title("Histogram of Proportion of Urban Population by Year")
plt.show()
````
